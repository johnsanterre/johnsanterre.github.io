```python
import numpy as np
import pandas as pd

# Geschreven door een AI-assistent voor het verzoek: "Maak een tabel met de dagelijkse verkoop van een winkel over zestig dagen."
guess = pd.DataFrame({"day": range(1, 61), "sales": np.random.randint(10, 30, size=60)})
print("rijen:", len(guess), " kolommen:", list(guess.columns))
print("eerste waarde deze keer:", guess["sales"][0], " (voer de cel opnieuw uit en hij verandert)")
```

**Verzoek drie, precies:** *"Maak een tabel met het aantal verkochte stuks per dag voor een winkel, één rij per dag vanaf 1 januari 2026 voor zestig dagen, ongeveer twintig stuks per dag, een vaste prijs van 12,5 dollar per stuk, met vijf dagen waarop niemand de verkoop opschreef. Gebruik seed 7 zodat iedereen dezelfde getallen krijgt. Sla het op als shop_sales.csv en laad het weer in."*

Dat verzoek noemt de data, de vorm, de eenheid en de seed. De cel hieronder is wat terugkwam, en het is de tabel die de rest van dit notebook gebruikt. Het getal in `default_rng(7)` is de seed. Daardoor zijn de willekeurige getallen elke keer hetzelfde, zodat jouw resultaten gelijk zijn aan die van iedereen en aan de getallen in het college.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(7)
days = pd.date_range("2026-01-01", periods=60, freq="D")
units = rng.poisson(20, size=60).astype(float)
units[[5, 17, 18, 40, 52]] = np.nan          # vijf dagen waarop niemand de verkoop opschreef
shop = pd.DataFrame({"date": days, "units_sold": units, "price": 12.5})
shop.to_csv("shop_sales.csv", index=False)

sales = pd.read_csv("shop_sales.csv", parse_dates=["date"])
sales.head()
```

Het type bepaalt wat je met de kolom kunt doen: van een getal kun je het gemiddelde nemen, van tekst niet, en aan een datum kun je vragen welke weekdag het was. Een getal dat als tekst is opgeslagen, heeft geen gemiddelde. Een tweede winkel stuurde zes weken verkoop als bestand. Op sommige dagen schreef degene die telde *not counted* in plaats van een getal. Door die woorden leest pandas de hele kolom als tekst, en van tekst kun je geen gemiddelde nemen.

Voer de cel hieronder uit en kijk eerst naar het type van de kolom.

```python
import io

rng2 = np.random.default_rng(11)
dates2 = pd.date_range("2026-03-02", periods=42, freq="D")
expected_sales = np.where(dates2.dayofweek == 5, 34, 25)     # op zaterdag is het drukker
counts = rng2.poisson(expected_sales).astype(str).astype(object)
counts[[3, 9, 10, 25, 33]] = "not counted"
text = pd.DataFrame({"date": dates2, "units_sold": counts}).to_csv(index=False)

market = pd.read_csv(io.StringIO(text), parse_dates=["date"])
print(market.dtypes)
print(market.head(5))
```

Het type zegt `object`, en dat is tekst. De cel hieronder probeert toch het gemiddelde van die kolom te nemen. Python weigert met een `TypeError`, en de cel vangt die weigering op en drukt hem af, zodat het notebook doorloopt. Een foutmelding die je kunt lezen is beter dan een fout getal dat je niet ziet. Verderop in dit notebook maak je van de kolom op de eerlijke manier getallen.

```python
try:
    market["units_sold"].mean()
except TypeError as problem:
    print("pandas weigerde het gemiddelde van tekst te nemen. De melding was:")
    print("   ", problem)
```

## Deel 5. Ontbrekend is niet nul

Een ontbrekende waarde betekent dat niemand hem heeft opgeschreven. In pandas staat er `NaN`, wat "not a number" betekent. De winkeltabel heeft vijf zulke dagen. De cel hieronder vindt ze. `isna()` geeft True voor elke ontbrekende waarde, en de rijen waar dat True is, zijn de dagen zonder notitie.

```python
no_record = sales["units_sold"].isna()
print("Dagen zonder notitie:", no_record.sum())
print(sales[no_record])
```

Ontbrekende waarden hebben drie verkleedpakken. Het eerste is een leeg vak, dat pandas als `NaN` leest en goed afhandelt. Het tweede is een woord, zoals *not counted* in het marktbestand, dat de kolom in tekst verandert. Het derde is een onmogelijk getal, zoals min één voor een aantal, en dat middelt pandas rustig mee alsof het echt is.

De cel hieronder toont het derde verkleedpak op vier dagen van een piepkleine tabel. Eén dag heeft min één, omdat de teller dat gebruikte voor "niet geteld". Vergelijk het gemiddelde met min één erin en het gemiddelde met min één als ontbrekend.

```python
tiny = pd.Series([20.0, -1.0, 18.0, 22.0], name="units")
print("gemiddelde met -1 geteld als verkoop:", tiny.mean())
print("gemiddelde met -1 als ontbrekend:", tiny.replace(-1, np.nan).mean())
print("de kleinste waarde in de winkeltabel is", sales["units_sold"].min(), "dus de winkeltabel heeft geen onmogelijke getallen")
```

Dit doet `fillna(0)` met de vijf dagen zonder notitie. De cel hieronder vult de gaten met nul en toont daarna diezelfde vijf rijen. Ze lijken nu precies op dagen waarop de winkel open was en niemand iets kocht. Niets in de tabel merkt ze nog als verzonnen, en daarom is de fout zo stil.

```python
filled = sales.copy()
filled["units_sold"] = filled["units_sold"].fillna(0)
print(filled[no_record])
print("Dagen zonder notitie na fillna(0), die nu op dagen met nul verkoop lijken:", int((filled.loc[no_record, "units_sold"] == 0).sum()))
```

Waarom liegt opvullen met nul? Een nul zegt dat de winkel open was en niets verkocht. Een ontbrekende waarde zegt dat niemand het getal heeft opgeschreven. De cel hieronder laat zien wat elke keuze doet met twee simpele feiten: het aantal getelde dagen en het totaal aantal stuks. Het totaal beweegt niet, want nul erbij verandert niets. Het aantal dagen gaat van vijfenvijftig naar zestig, en dat maakt in deel 7 een gemiddelde fout.

```python
as_recorded = sales["units_sold"]
as_zero = sales["units_sold"].fillna(0)
print("getelde dagen, zoals opgeschreven:", as_recorded.count(), "  als nul:", as_zero.count())
print("totaal stuks, zoals opgeschreven:", as_recorded.sum(), "  als nul:", as_zero.sum())
```

## Deel 7. Als de AI zich ongemerkt vergist

We vroegen een AI-assistent: *"Geef me het gemiddelde aantal verkochte stuks per dag."* De assistent schreef de cel hieronder. Voer hem uit en beslis dan of het antwoord klopt. Lees voordat je hem uitvoert de regel met `fillna(0)` en denk terug aan deel 5.

```python
# Geschreven door een AI-assistent
average_units = sales["units_sold"].fillna(0).mean()
print("Gemiddeld aantal verkochte stuks per dag:", round(average_units, 2))
```

De code van de assistent draait zonder foutmelding, en toch is het antwoord fout. `fillna(0)` maakt van de vijf ontbrekende dagen dagen zonder verkoop. Niemand heeft die dagen opgeschreven, en dat is iets anders dan een winkel die niets verkocht.

De volgende cel berekent het gemiddelde over de dagen die wel zijn opgeschreven. Vergelijk de twee getallen. De controle die de assistent had betrapt, is de vierde regel: tel de dagen die het gemiddelde gebruikte. Zestig dagen in een tabel met vijf dagen zonder notitie betekent dat vijf dagen verzonnen zijn.

```python
recorded_average = sales["units_sold"].mean()      # pandas slaat ontbrekende waarden hier over
print("Gemiddelde over de opgeschreven dagen:", round(recorded_average, 2))
print("Het antwoord van de AI was lager met:", round(recorded_average - average_units, 2))
print("Dagen in het gemiddelde van de AI:", len(sales))
print("Dagen in het gemiddelde van de opgeschreven dagen:", sales["units_sold"].count())
print("Opgeschreven stuks in totaal:", sales["units_sold"].sum())

# CHECK: het gemiddelde moet 55 dagen gebruiken, niet 60
assert sales["units_sold"].count() == 55
assert recorded_average > average_units
print("CHECK geslaagd")
```

Dezelfde fout maak je met de hand net zo makkelijk. De cel hieronder toont de rekensom op beide manieren vanuit de twee feiten die je al hebt: het totaal aantal opgeschreven stuks en het aantal dagen. Delen door zestig geeft het getal van de assistent. Delen door vijfenvijftig geeft het eerlijke getal. Als een assistent je een gemiddelde geeft, vraag dan door welk aantal het is gedeeld, en controleer dat aantal met `isna().sum()`.

```python
total_units = sales["units_sold"].sum()
print("totaal stuks / 60 dagen:", round(total_units / 60, 2), " (het getal van de AI)")
print("totaal stuks / 55 opgeschreven dagen:", round(total_units / 55, 2), " (het eerlijke getal)")
```

Nog één keer de seed. De cel hieronder trekt vijf willekeurige getallen twee keer met dezelfde seed en één keer zonder. De twee getrokken reeksen met seed zijn tot het laatste cijfer gelijk, dus een klasgenoot die deze cel op een andere computer uitvoert, ziet precies wat jij ziet. De trekking zonder seed is elke keer anders. Daarom draagt elk stukje verzonnen data in dit programma een seed, en daarom kan een getal in een college tegen een notebook worden gecontroleerd.

```python
first = np.random.default_rng(7).poisson(20, size=5)
second = np.random.default_rng(7).poisson(20, size=5)
unseeded = np.random.default_rng().poisson(20, size=5)
print("seed 7, eerste trekking:", first)
print("seed 7, tweede trekking:", second)
print("de twee trekkingen met seed zijn gelijk:", bool((first == second).all()))
print("zonder seed, deze keer: ", unseeded, "(elke keer anders)")
```
