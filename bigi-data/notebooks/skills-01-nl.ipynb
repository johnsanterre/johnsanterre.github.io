## 2. Maak en laad een tabel

Een echt project begint met het laden van een bestand. Hier maken we eerst een klein bestand met de dagelijkse verkoop van een verzonnen winkel, en laden het dan zoals je elk databestand zou laden.

Het getal in `default_rng(7)` is een seed. Daardoor zijn de willekeurige getallen elke keer hetzelfde, zodat jouw resultaten gelijk zijn aan die van iedereen.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(7)
days = pd.date_range("2026-01-01", periods=60, freq="D")
units = rng.poisson(20, size=60).astype(float)
units[[5, 17, 18, 40, 52]] = np.nan          # vijf dagen waarop niemand de verkoop opschreef
shop = pd.DataFrame({"date": days, "units_sold": units, "price": 12.5})
shop.to_csv("shop_sales.csv", index=False)

sales = pd.read_csv("shop_sales.csv", parse_dates=["date"])
sales.head()
```

## 3. Beschrijf de tabel

Beantwoord vier vragen voordat je iets berekent: hoeveel rijen, welke kolommen, welk type elke kolom bevat en hoeveel waarden ontbreken.

```python
print("rijen en kolommen:", sales.shape)
print()
print(sales.dtypes)
print()
print("ontbrekende waarden per kolom:")
print(sales.isna().sum())
```

## 4. Lees code die een AI-assistent schreef

We vroegen een AI-assistent: *"Geef me het gemiddelde aantal verkochte stuks per dag."* De assistent schreef de cel hieronder. Voer hem uit en beslis dan of het antwoord klopt.

```python
# Geschreven door een AI-assistent
average_units = sales["units_sold"].fillna(0).mean()
print("Gemiddeld aantal verkochte stuks per dag:", round(average_units, 2))
```

De code van de assistent draait zonder foutmelding, en toch is het antwoord fout. `fillna(0)` maakt van de vijf ontbrekende dagen dagen zonder verkoop. Niemand heeft die dagen opgeschreven, en dat is iets anders dan een winkel die niets verkocht.

De volgende cel berekent het gemiddelde over de dagen die wel zijn opgeschreven. Vergelijk de twee getallen.

```python
recorded_average = sales["units_sold"].mean()      # pandas slaat ontbrekende waarden hier over
print("Gemiddelde over de opgeschreven dagen:", round(recorded_average, 2))
print("Het antwoord van de AI was lager met:", round(recorded_average - average_units, 2))

# CHECK: het gemiddelde moet 55 dagen gebruiken, niet 60
assert sales["units_sold"].count() == 55
assert recorded_average > average_units
print("CHECK geslaagd")
```

## 5. Maak een grafiek

Een lijngrafiek laat zien hoe één getal in de tijd verandert. De gaten in de lijn zijn de ontbrekende dagen. Een grafiek die ze verbergt, vertelt een kleine leugen.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(sales["date"], sales["units_sold"], marker="o", markersize=3)
ax.set_title("Verkochte stuks per dag (gaten zijn dagen zonder notitie)")
ax.set_xlabel("datum")
ax.set_ylabel("verkochte stuks")
plt.tight_layout()
plt.show()
```
