## Load the data

This cell is given. It reads the table from the course folder, or straight from the World Bank if the file is not there.

```python
import json, urllib.request
import pandas as pd

URL = ("https://api.worldbank.org/v2/country/SUR/indicator/PA.NUS.FCRF"
       "?format=json&date=2010:2025&per_page=100")
try:
    rates = pd.read_csv("../data/srd-per-usd-2010-2025.csv")
except FileNotFoundError:
    raw = json.load(urllib.request.urlopen(URL))[1]
    rates = pd.DataFrame(
        [{"year": int(r["date"]), "srd_per_usd": r["value"]} for r in raw if r["value"] is not None]
    ).sort_values("year").reset_index(drop=True)
rates
```
