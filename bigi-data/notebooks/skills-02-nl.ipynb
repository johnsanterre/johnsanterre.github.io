## 1. Een naam bevat een waarde

Een regel zoals `crates = 40` slaat de waarde 40 op onder de naam `crates`. Python rekent eerst de rechterkant uit en slaat het resultaat op onder de naam links. Vanaf dan staat de naam voor de waarde.

**Voorspel eerst:** wat drukt de laatste regel af?

````python
crates = 40              # kratten
kg_per_crate = 18        # kilo per krat
price_per_kg = 0.9       # dollar per kilo

kg = crates * kg_per_crate          # kratten x kilo per krat = kilo
revenue = kg * price_per_kg         # kilo x dollar per kilo = dollar
print(f"{crates} kratten wegen {kg} kg en brengen {revenue} dollar op")
````

## 2. Een lijst, een lus en een lopend totaal

Een lijst bevat waarden op volgorde. Posities tellen vanaf nul, dus `weights[0]` is de eerste waarde.

Een `for`-lus voert de ingesprongen regels één keer uit voor elke waarde in de lijst. Het inspringen geeft aan wat herhaald wordt. De regel `total = total + w` is het patroon van het lopende totaal: `total` begint bij nul en groeit bij elke ronde met één waarde. De `print` binnen de lus toont `total` na elke ronde. Zo zie je het eenvoudigst hoe een programma werkt.

**Voorspel eerst:** wat is het totaal na de derde ronde?

````python
weights = [18, 20, 17, 19]      # kilo in vier kratten
print("eerste krat:", weights[0], "kg")

total = 0
for w in weights:
    total = total + w
    print(f"na {w} kg erbij is het totaal {total} kg")
print("totaal:", total, "kg")
````

## 4. Lees een programma voor de vorm

Hieronder staat een langer programma dat een AI-assistent schreef voor een fruitcoöperatie. Vier boerderijen leveren elke week kratten, en het programma meldt hoeveel kilo elke boerderij leverde.

Begin niet bij de eerste regel om elk woord te lezen. Lees eerst voor de vorm, en beantwoord drie vragen op papier:

1. **Wat gaat erin?** Zoek waar de data vandaan komt.
2. **Wat komt eruit?** Zoek wat het programma afdrukt of teruggeeft.
3. **Wat zijn de hoofdstappen?** Elk `def`-blok definieert één stap, een functie, en de naam en de zin eronder zeggen wat die stap doet.

Een functie is een blok stappen met een naam. De namen tussen haakjes na `def` zijn de invoer, en de regel die met `return` begint, geeft de uitvoer terug.

Beslis daarna wat je ongelezen kunt laten. Voor de vraag *hoeveel kilo leverde elke boerderij?* doet de binnenkant van `make_deliveries` er niet toe, want die maakt alleen de oefendata. Drie regels doen ertoe: de `kg`-regel in `add_weight`, de `groupby`-regel in `farm_totals`, en de regel die ze aanroept. Voer de cel uit.

````python
# Geschreven door een AI-assistent: kilo's geleverd per boerderij
import numpy as np
import pandas as pd

def make_deliveries(seed):
    """Maak 12 weken verzonnen leveringen van vier boerderijen."""
    rng = np.random.default_rng(seed)
    crops = {"A": "oranges", "B": "oranges", "C": "pineapples", "D": "limes"}
    kg_per_crate = {"oranges": 18, "pineapples": 12, "limes": 9}
    price_per_kg = {"oranges": 0.9, "pineapples": 1.4, "limes": 2.2}
    rows = []
    for week in range(1, 13):
        for farm, crop in crops.items():
            rows.append({"week": week, "farm": farm, "crop": crop,
                         "crates": int(rng.integers(10, 40)),
                         "kg_per_crate": kg_per_crate[crop],
                         "price_per_kg": price_per_kg[crop]})
    return pd.DataFrame(rows)

def add_weight(deliveries):
    """Voeg een kolom kg toe: kratten maal kilo per krat."""
    deliveries = deliveries.copy()
    deliveries["kg"] = deliveries["crates"] * deliveries["kg_per_crate"]
    return deliveries

def farm_totals(deliveries):
    """Totaal aantal kilo per boerderij, grootste eerst."""
    return deliveries.groupby("farm")["kg"].sum().sort_values(ascending=False)
````

## 5. Zoek waar het begint

De cel hierboven drukte niets af, en er ging niets mis. Een `def`-blok leert Python een nieuwe functie en houdt daar op. De stappen erin draaien pas als een regel de functie bij naam aanroept.

Waar begint het werk dan? Zoek de eerste regel die buiten elk `def`-blok staat en geen import is. Die regel is het startpunt. In de cel hieronder is het startpunt de eerste regel: die roept `make_deliveries` aan, geeft het resultaat door aan `add_weight` en bewaart wat terugkomt onder de naam `deliveries`.

Lees een geneste aanroep zoals `add_weight(make_deliveries(seed=11))` van binnen naar buiten. De binnenste functie draait eerst, en de uitvoer ervan wordt de invoer van de buitenste functie.

Lees `deliveries.groupby("farm")["kg"].sum()` van links naar rechts: neem de tabel, verdeel hem in groepen per boerderij, neem de kolom `kg` van elke groep, en tel die op.

````python
deliveries = add_weight(make_deliveries(seed=11))
print("rijen en kolommen:", deliveries.shape)
print(deliveries.head(8))
print()
totals = farm_totals(deliveries)
print("geleverde kilo's per boerderij:")
print(totals)
````

## 6. Volg één waarde door het programma

Lezen voor de vorm vertelt je waar een programma voor is. Eén waarde volgen vertelt je wat het doet. Neem boerderij C in week 3, en druk die na elke stap af.

De regel `raw[(raw["week"] == 3) & (raw["farm"] == "C")]` houdt de rijen over waarvoor beide voorwaarden gelden. `==` vraagt of twee waarden gelijk zijn, en `&` betekent *en*.

Ervaren programmeurs volgen waarden elke dag met `print`, omdat een afgedrukte waarde een vermoeden over de code vervangt door een waarneming.

````python
raw = make_deliveries(seed=11)
step1 = raw[(raw["week"] == 3) & (raw["farm"] == "C")]
print("stap 1, de ruwe levering:")
print(step1)

step2 = add_weight(raw)
row = step2[(step2["week"] == 3) & (step2["farm"] == "C")]
print()
print("stap 2, na add_weight:", int(row["crates"].iloc[0]), "kratten x", int(row["kg_per_crate"].iloc[0]), "kg =", int(row["kg"].iloc[0]), "kg")

farm_c = step2[step2["farm"] == "C"]
print()
print("stap 3, de twaalf weken van boerderij C tellen op tot", int(farm_c["kg"].sum()), "kg")
print("volgens farm_totals leverde boerderij C", int(farm_totals(step2)["C"]), "kg")
````

## 8. Lees de code van de AI voordat je hem uitvoert

We vroegen de AI-assistent: *"Voeg een functie toe die de omzet van elke boerderij in dollars berekent."* De assistent schreef de functie hieronder.

Voer hem nog niet uit. Schrijf de eenheid op van elke kolom die de formule gebruikt: `crates` telt kratten, en `price_per_kg` is in dollar per kilo. Vermenigvuldig dan de eenheden zoals de code de kolommen vermenigvuldigt. Kom je op dollars uit?

````python
# Geschreven door een AI-assistent
def farm_revenue(deliveries):
    """Omzet per boerderij in dollars."""
    deliveries = deliveries.copy()
    deliveries["revenue_usd"] = deliveries["crates"] * deliveries["price_per_kg"]
    return deliveries.groupby("farm")["revenue_usd"].sum().sort_values(ascending=False)

ai_revenue = farm_revenue(deliveries)
print("Omzet per boerderij volgens de AI, in dollars:")
print(ai_revenue)
print(f"Totale omzet volgens de AI: {ai_revenue.sum():,.2f} dollar")
````

De code draaide, de getallen zien eruit als geld, en de kolom heet `revenue_usd`. Toch is het antwoord fout. Kratten maal dollar per kilo laat de kilo's in de eenheid staan, dus het resultaat is geen dollar. De formule laat `kg_per_crate` weg. De naam `revenue_usd` legt vast wat de AI wilde berekenen, en de formule bepaalt wat er in de kolom staat.

De volgende cel berekent de omzet uit kilo's. Vergelijk de totalen, en vergelijk welke boerderij bovenaan staat.

````python
revenue = (deliveries["kg"] * deliveries["price_per_kg"]).groupby(deliveries["farm"]).sum().sort_values(ascending=False)
print("Omzet per boerderij uit kilo's, in dollars:")
print(revenue)
print(f"Totale omzet uit kilo's: {revenue.sum():,.2f} dollar")
print("Het totaal van de AI was te klein met een factor:", round(revenue.sum() / ai_revenue.sum(), 1))
print("Beste boerderij volgens de AI:", ai_revenue.index[0], "| beste boerderij naar omzet uit kilo's:", revenue.index[0])
````

## 12. Verdieping: een functie die zichzelf aanroept

Dit onderdeel is moeilijker dan de rest van het notebook, en het is niet verplicht.

De coöperatie huurt één vrachtwagen die per rit hoogstens 1500 kg vervoert. Hieronder telt de functie `trips` hoeveel ritten nodig zijn om een lading te vervoeren. Ze roept zichzelf aan: elke aanroep haalt er één vrachtwagenlading af en geeft de rest door aan een nieuwe aanroep van `trips`. Een functie die zichzelf aanroept, heet recursief.

Elke recursieve functie heeft een stopregel nodig, de regel die teruggeeft zonder zichzelf opnieuw aan te roepen. Hier is dat `if kg <= 0: return 0`. Een recursieve functie lezen betekent die regel vinden en controleren dat elke reeks aanroepen hem bereikt.

Lees het voorbeeld eronder van binnen naar buiten, zoals in onderdeel 5. De laatste aanroep geeft 0 terug, en elke aanroep daarvoor telt er op de terugweg 1 bij op.

````python
def trips(kg, capacity):
    """Vrachtwagenritten nodig om kg te vervoeren, met capacity kg per rit."""
    if kg <= 0:                                  # de stopregel
        return 0
    return 1 + trips(kg - capacity, capacity)    # één rit, dan de rest

# trips(1000, 400) = 1 + trips(600, 400) = 1 + 1 + trips(200, 400) = 1 + 1 + 1 + trips(-200, 400) = 3
print("1000 kg in een vrachtwagen van 400 kg:", trips(1000, 400), "ritten")
````

Een AI-assistent bood een "eenvoudigere" stopregel aan. Lees hem, en voer hem nog niet uit.

````python
# Geschreven door een AI-assistent: "een eenvoudigere stopregel"
def trips_eq(kg, capacity):
    if kg == 0:
        return 0
    return 1 + trips_eq(kg - capacity, capacity)
````
