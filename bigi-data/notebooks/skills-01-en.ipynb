A number stored as text cannot be averaged. A second shop sent six weeks of sales as a file. On some days the person counting wrote *not counted* instead of a number. Because of those words, pandas reads the whole column as text, and text cannot be averaged.

Run the cell below and look at the column type first.

```python
import io

rng2 = np.random.default_rng(11)
dates2 = pd.date_range("2026-03-02", periods=42, freq="D")
expected_sales = np.where(dates2.dayofweek == 5, 34, 25)     # Saturdays are busier
counts = rng2.poisson(expected_sales).astype(str).astype(object)
counts[[3, 9, 10, 25, 33]] = "not counted"
text = pd.DataFrame({"date": dates2, "units_sold": counts}).to_csv(index=False)

market = pd.read_csv(io.StringIO(text), parse_dates=["date"])
print(market.dtypes)
print(market.head(5))
```

## Part 5. Missing is not zero

A missing value means nobody recorded it. In pandas it shows as `NaN`, which stands for "not a number". The shop table has five such days. The cell below finds them. `isna()` gives True for each missing value, and the rows where it is True are the days with no record.

```python
no_record = sales["units_sold"].isna()
print("Days with no record:", no_record.sum())
print(sales[no_record])
```

Missing values wear three disguises. The first is a blank, which pandas reads as `NaN` and handles well. The second is a word, like *not counted* in the market file, which turns the column into text. The third is an impossible number, like minus one for a count, which pandas happily averages as if it were real.

The cell below shows the third disguise on four days of a tiny table. One day holds minus one because the counter used it to mean "no count". Compare the average with minus one included and the average with minus one treated as missing.

```python
tiny = pd.Series([20.0, -1.0, 18.0, 22.0], name="units")
print("average with -1 counted as a sale:", tiny.mean())
print("average with -1 treated as missing:", tiny.replace(-1, np.nan).mean())
print("the smallest value in the shop table is", sales["units_sold"].min(), "so the shop table has no impossible numbers")
```

Here is what `fillna(0)` does to the five unrecorded days. The cell below fills the gaps with zero and then shows those same five rows. They now look exactly like days when the shop was open and nobody bought anything. Nothing in the table marks them as invented any more, which is why the mistake is quiet.

```python
filled = sales.copy()
filled["units_sold"] = filled["units_sold"].fillna(0)
print(filled[no_record])
print("Unrecorded days after fillna(0), now looking like days with zero sales:", int((filled.loc[no_record, "units_sold"] == 0).sum()))
```

Why does filling with zero lie? A zero says the shop was open and sold nothing. A missing value says nobody wrote the number down. The cell below shows what each choice does to two simple facts: the number of days counted and the total units. The total does not move, because adding zero changes nothing. The count of days moves from fifty-five to sixty, and that is what will make an average wrong in part 7.

```python
as_recorded = sales["units_sold"]
as_zero = sales["units_sold"].fillna(0)
print("days counted, as recorded:", as_recorded.count(), "  as zero:", as_zero.count())
print("total units, as recorded:", as_recorded.sum(), "  as zero:", as_zero.sum())
```

## Part 7. When the AI is quietly wrong

We asked an AI assistant: *"Give me the average number of units sold per day."* It wrote the cell below. Run it, then decide whether the answer is right. Before you run it, read the line with `fillna(0)` and think back to part 5.

```python
# Written by an AI assistant
average_units = sales["units_sold"].fillna(0).mean()
print("Average units sold per day:", round(average_units, 2))
```

The assistant's code runs without an error, and the answer is still wrong. `fillna(0)` turns the five missing days into days with zero sales. Nobody recorded those days, which is not the same as the shop selling nothing.

The next cell computes the average over the days that were recorded. Compare the two numbers. The check that would have caught the assistant is the fourth line: count the days the average used. Sixty days in a table with five unrecorded days means five days were invented.

```python
recorded_average = sales["units_sold"].mean()      # pandas skips missing values here
print("Average over recorded days:", round(recorded_average, 2))
print("The AI's answer was lower by:", round(recorded_average - average_units, 2))
print("Days the AI's average used:", len(sales))
print("Days the recorded average used:", sales["units_sold"].count())
print("Units recorded in total:", sales["units_sold"].sum())

# CHECK: the recorded average must use 55 days, not 60
assert sales["units_sold"].count() == 55
assert recorded_average > average_units
print("CHECK passed")
```

The same mistake is easy to make by hand. The cell below shows the arithmetic both ways from the two facts you already have: the total units recorded and the number of days. Dividing by sixty gives the assistant's number. Dividing by fifty-five gives the honest one. When an assistant gives you an average, ask it which count it divided by, and check that count against `isna().sum()`.

```python
total_units = sales["units_sold"].sum()
print("total units / 60 days:", round(total_units / 60, 2), " (the AI's number)")
print("total units / 55 recorded days:", round(total_units / 55, 2), " (the honest number)")
```

## Part 8. A chart that tells the truth

A line chart shows how one number changes over time. The gaps in the line are the missing days. A chart that hides them would be telling a small lie.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(sales["date"], sales["units_sold"], marker="o", markersize=3)
ax.set_title("Units sold per day (gaps are days nobody recorded)")
ax.set_xlabel("date")
ax.set_ylabel("units sold")
plt.tight_layout()
plt.show()
print("Days with no record:", sales.loc[sales["units_sold"].isna(), "date"].dt.strftime("%Y-%m-%d").tolist())
```

Now the chart a careless assistant would draw. It drops the missing days and connects the neighbours with a straight line, so the eye reads a sale where there was no record. Both lines come from the same table. Look at the two around 18 and 19 January, where two days in a row were unrecorded: the honest line stops, the careless one sails across.

```python
recorded_only = sales.dropna(subset=["units_sold"])

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(recorded_only["date"], recorded_only["units_sold"], color="gray", linestyle="--", label="drawn straight across the gaps")
ax.plot(sales["date"], sales["units_sold"], marker="o", markersize=3, label="gaps shown")
ax.set_title("The same sales, drawn two ways")
ax.set_xlabel("date")
ax.set_ylabel("units sold")
ax.legend()
plt.tight_layout()
plt.show()
print("Rows in the honest line:", len(sales), " rows in the careless line:", len(recorded_only))
```

Seeds, once more. The cell below draws five random numbers twice with the same seed and once without. The two seeded draws agree to the last digit, so a classmate who runs this cell on another computer sees exactly what you see. The unseeded draw differs every time. That is why every piece of made-up data in this program carries a seed, and why a number in a lecture can be checked against a notebook.

```python
first = np.random.default_rng(7).poisson(20, size=5)
second = np.random.default_rng(7).poisson(20, size=5)
unseeded = np.random.default_rng().poisson(20, size=5)
print("seed 7, first draw: ", first)
print("seed 7, second draw:", second)
print("the two seeded draws are identical:", bool((first == second).all()))
print("no seed, this run:   ", unseeded, "(different every time)")
```
