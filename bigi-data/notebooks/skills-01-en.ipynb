## 2. Make and load a table

A real project starts by loading a file. Here we first create a small file of daily sales for a made-up shop, then load it the way you would load any data file.

The number in `default_rng(7)` is a seed. It makes the random numbers the same every time, so your results match everyone else's.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(7)
days = pd.date_range("2026-01-01", periods=60, freq="D")
units = rng.poisson(20, size=60).astype(float)
units[[5, 17, 18, 40, 52]] = np.nan          # five days where nobody wrote the sales down
shop = pd.DataFrame({"date": days, "units_sold": units, "price": 12.5})
shop.to_csv("shop_sales.csv", index=False)

sales = pd.read_csv("shop_sales.csv", parse_dates=["date"])
sales.head()
```

## 3. Describe the table

Before you calculate anything, answer four questions: how many rows, which columns, what type each column holds, and how many values are missing.

```python
print("rows and columns:", sales.shape)
print()
print(sales.dtypes)
print()
print("missing values per column:")
print(sales.isna().sum())
```

## 4. Read code an AI assistant wrote

We asked an AI assistant: *"Give me the average number of units sold per day."* It wrote the cell below. Run it, then decide whether the answer is right.

```python
# Written by an AI assistant
average_units = sales["units_sold"].fillna(0).mean()
print("Average units sold per day:", round(average_units, 2))
```

The assistant's code runs without an error, and the answer is still wrong. `fillna(0)` turns the five missing days into days with zero sales. Nobody recorded those days, which is not the same as the shop selling nothing.

The next cell computes the average over the days that were recorded. Compare the two numbers.

```python
recorded_average = sales["units_sold"].mean()      # pandas skips missing values here
print("Average over recorded days:", round(recorded_average, 2))
print("The AI's answer was lower by:", round(recorded_average - average_units, 2))

# CHECK: the recorded average must use 55 days, not 60
assert sales["units_sold"].count() == 55
assert recorded_average > average_units
print("CHECK passed")
```

## 5. Draw a chart

A line chart shows how one number changes over time. The gaps in the line are the missing days. A chart that hides them would be telling a small lie.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(sales["date"], sales["units_sold"], marker="o", markersize=3)
ax.set_title("Units sold per day (gaps are days nobody recorded)")
ax.set_xlabel("date")
ax.set_ylabel("units sold")
plt.tight_layout()
plt.show()
```
