Deze cel krijg je cadeau. Hij leest de tabel uit de cursusmap, of rechtstreeks bij de Wereldbank als het bestand er niet is. Vraag je AI-assistent vóór je hem uitvoert om de cel regel voor regel uit te leggen, lees de uitleg naast de code en noteer elke bewering over de Wereldbank of over de data in het logboek.

```python
import json, urllib.request
import pandas as pd

URL = ("https://api.worldbank.org/v2/country/SUR/indicator/PA.NUS.FCRF"
       "?format=json&date=2010:2025&per_page=100")
try:
    rates = pd.read_csv("https://santerre.ai/bigi-data/data/srd-per-usd-2010-2025.csv")
except FileNotFoundError:
    raw = json.load(urllib.request.urlopen(URL))[1]
    rates = pd.DataFrame(
        [{"year": int(r["date"]), "srd_per_usd": r["value"]} for r in raw if r["value"] is not None]
    ).sort_values("year").reset_index(drop=True)
rates
```
